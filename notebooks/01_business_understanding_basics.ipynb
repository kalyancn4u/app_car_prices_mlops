**Phase goal:** describe the problem in *business* terms and write down the
measurable targets (KPIs) a model must hit before anyone is allowed to ship it.

## The problem
A used-car marketplace wants an **instant, trustworthy price estimate** for any
car a seller lists, plus a simple **budget band** (Low / Medium / High) for
search filters. Hand appraisal is slow and inconsistent; a model that is *fast,
accurate and explainable* directly improves listings and pricing confidence.

## Who cares, and about what?
| Stakeholder | What they need | The number that measures it |
| :-- | :-- | :-- |
| Seller | A fair, believable price | Typical error in ₹ (**MAE**) |
| Buyer | Sensible budget bands | **Band accuracy** |
| Product | Confidence to auto-suggest a price | **R²** (variance explained) |
| Engineering | Something cheap and easy to serve | A small, fast model file |

## The machine-learning task
We predict the **exact price in ₹ Lakhs** with a **regression** model. (1 Lakh =
₹100,000, so `4.75` means ₹4,75,000.) The **Low/Medium/High band is *derived*
from the predicted price** — we do *not* train a second classifier. That way the
band and the rupee figure can never contradict each other.

```python
# The whole project is governed by a handful of constants. In the real codebase
# these live in src/car_pricing/config.py; here we simply write them out so this
# notebook depends on nothing but the Python standard library.

# The business "ship / no-ship" gate: a model must clear ALL of these.
KPI = {
    "max_mae_lakhs":     1.0,   # typical error must be under Rs 1,00,000
    "min_r2":            0.85,  # explain at least 85% of the price variance
    "min_band_accuracy": 0.70,  # correct budget band at least 70% of the time
}

PRICE_UNIT  = "Lakhs"
BAND_LABELS = ["Low", "Medium", "High"]

print("A candidate model may ship only if it clears every KPI gate:")
for name, value in KPI.items():
    print(f"  {name:20s} {value}")

print("\nPrice unit:", PRICE_UNIT, "| Budget bands:", BAND_LABELS)
```

## Success criteria (the gate, in plain English)
The training pipeline **automatically refuses to ship** a model that fails any of:

- **MAE ≤ ₹1.0 Lakh** — the typical miss must be smaller than ₹1,00,000.
- **R² ≥ 0.85** — explain at least 85 % of the variation in price.
- **Band accuracy ≥ 70 %** — put the car in the right budget bucket most of the time.

> **Why these three and not just one?** R² rewards explaining the *overall*
> spread but is easy to inflate; MAE is the number a seller actually feels (“how
> many rupees off, typically?”); band accuracy checks the product feature we
> ship. A good model has to satisfy all three at once.
